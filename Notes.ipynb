```python

import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Category': ['A', 'A', 'B', 'B', 'A', 'B','A','B'],
    'Value': [10, 15, 20, 25, 30, 35,40,50],
    'Fee' :[20000,25000,26000,22000,24000,35000,4000,5600],
    'Duration':['30day','40days','35days','40days','60days','60days','50days','60days'],
    'Discount':[1000,2300,1200,2500,2000,2000,500,600]
})
df
# Aggregate Functions in Pandas
# calculate the sum of the 'Value' column using agg(function)
total_sum=df[['Value']].agg('sum')
print(total_sum)
# another way to calculate the sum
total_sum=df[['Value']].sum()
print(total_sum)
# calculating cumulative sum
total_cumsum=df[['Value']].cumsum()
print(total_cumsum)
# another way to calculate the cumulative method
total_cumsum=df[['Value']].cummax()
print(total_cumsum)
# another way to calculate the cumulative method
total_cumsum=df[['Value']].cummin()
print(total_cumsum)

# calculate the mean of the 'Value' column
total_mean=df[['Value']].agg('mean')
print(total_mean)

# calculate the maximum of the 'Value' column
total_max=  df[['Value']].agg('max')
print(total_max)

# calculate the minimum of the 'Value' column
total_min=df[['Value']].agg('min')
print(total_min)
# calculate multiple aggregations at once
total_agg=df[['Discount', 'Fee']].agg(['sum', 'mean', 'max', 'min'])
print(total_agg)
# group by 'Category' and calculate aggregations
grouped_agg=df.groupby('Category')[['Value','Discount', 'Fee']].agg(['sum', 'mean', 'max', 'min'])
print(grouped_agg)
```
